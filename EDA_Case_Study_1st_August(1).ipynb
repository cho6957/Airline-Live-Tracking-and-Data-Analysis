Exploratory Data Analysis (EDA) is the process of examining, understanding, and summarizing data before applying Machine Learning or Statistical Models.
EDA helps identify patterns, trends, missing values, outliers, and relationships within data.

Exploratory Data Analysis (EDA) is the process of analyzing datasets using statistical and visualization techniques to understand the structure and quality of the data.

Objectives of EDA
• Understand the dataset. • Identify missing values. • Detect outliers. • Find patterns and trends. • Discover relationships between variables. • Prepare data for Machine Learning.

Why is EDA Important?
• Improves data quality. • Helps make better decisions. • Reduces modeling errors. • Reveals hidden insights. • Increases model accuracy.

Applications of EDA

• Data Science • Machine Learning • Business Analytics • Market Research • Financial Analysis • Healthcare Analytics

#apply replace for below column
replace ? with nan (Not a Number) or desired value which conveys information

```python
data["workclass"] = data.workclass.str.strip()
```

```python
data['workclass'] = data.workclass.replace('?',np.nan)
```

```python
data['workclass'].unique()
```

```python
#Handle the extra spaces
data['education'] = data['education'].str.strip()
```

```python
#same way we can check for other columns
data['occupation'] = data['occupation'].str.replace('?','other')
data['occupation'].unique()
```

```python
data['native.country'] = data['native.country'].str.replace('?','notavail')
data['native.country'].unique()
```

```python
#we want to remove all extra in all object columns at a time
#Anonymous functions --> lambda params : expression
data = data.apply(lambda l : l.str.strip() if l.dtype == 'object' else l)
```

```python
#Removing spaces together is very important to understand about the data
data.select_dtypes(include = np.int64)
```

```python
print(data['hours.per.week'].unique())
print(len(data['hours.per.week'].unique()))
```

```python
print(data['age'].unique())
print(len(data['age'].unique()))
```
